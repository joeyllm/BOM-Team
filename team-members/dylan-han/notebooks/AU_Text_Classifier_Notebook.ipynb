All of these directly show that the article is discussing the Australian market.

## Heuristic Thinking for Identifying Australian-Local Text

We can divide the judgment into three levels.

### Level 1: Strong Signals

If these appear, the text is very likely to be Australian-local.

#### 1. Australian domains

For example, the URL contains:

- `.com.au`
- `.net.au`
- `abc.net.au`
- `smh.com.au`
- `theage.com.au`
- `canberratimes.com.au`
- `brisbanetimes.com.au`
- `news.com.au`
- `afr.com`

#### 2. Australian institutions, places, and markets

For example, the text contains:

- `Australia`, `Australian`
- `ASX`
- `Canberra`, `Sydney`, `Melbourne`, `Brisbane`, `Perth`, `Adelaide`
- `NSW`, `VIC`, `QLD`, `WA`, `SA`, `ACT`, `Tasmania`
- `Parliament House`
- `ATO`, `Centrelink`, `Medicare`
- `Treasurer`, `Prime Minister` when the person mentioned is an Australian political figure
- Australian company names such as `Telstra`, `BHP`, `NAB`, `Commonwealth Bank`, `Woolworths`, `Coles`, and `Qantas`

---

### Level 2: Medium Signals

#### 3. Australian spelling

For example:

- `organisation` instead of `organization`
- `labour` instead of `labor`
- `centre` instead of `center`
- `favour`, `colour`, `behaviour`
- `realise`, `recognise`

These spellings are closer to British-style English. They are common in Australian English, but they are also common in British English, so they should not be used as the only deciding evidence.

#### 4. Australian local vocabulary

For example:

- `petrol`
- `mum`
- `uni`
- `footpath`
- `suburb`
- `holiday house`
- `bushfire`
- `bottle shop`
- `esky`

---

### Level 3: Exclusion Signals

If the text is clearly in an American or British local context, it should not be classified as Australian-local.

#### Strong American signals

- `.com` American media sources such as CNN, NYT, Fox, or The Washington Post
- `US Congress`, `Senate`, `White House`
- `dollars` in a clearly US economic context
- `high school`, `mom`, `gas station`, `apartment`, `downtown`
- `S&P 500`, `NASDAQ`, `Wall Street` when the whole text is centered on US local events

#### Strong British signals

- `.co.uk`, `bbc.co.uk`, `theguardian.com/uk`
- `Parliament` in a British context
- `NHS`
- `MP`, `Downing Street`
- `pounds`, `Bank of England`
- `Manchester`, `London`, `Scotland`, `Wales`

```python
import pandas as pd

# Assume this is the DataFrame you have already loaded
# If needed, uncomment the next line to read the parquet file
# df = pd.read_parquet("~/data/AUTokens50/part_x.parquet")

# Select one record for validation
row = df.loc[9659]

text = str(row["text"])
url = str(row["url"])

text_lower = text.lower()
url_lower = url.lower()

# -------------------------
# Define scoring rules
# -------------------------

australian_domains = [
    ".com.au", ".net.au", "abc.net.au", "smh.com.au", "theage.com.au",
    "canberratimes.com.au", "brisbanetimes.com.au", "news.com.au", "afr.com"
]

australian_keywords = [
    "australia", "australian", "asx", "canberra", "sydney", "melbourne",
    "brisbane", "perth", "adelaide", "tasmania", "new south wales",
    "queensland", "victoria", "western australia", "south australia",
    "nsw", "vic", "qld", "wa", "sa", "act",
    "parliament house", "ato", "centrelink", "medicare",
    "telstra", "bhp", "nab", "commonwealth bank", "woolworths", "coles", "qantas"
]

australian_spellings = [
    "organisation", "labour", "centre", "favour", "colour",
    "behaviour", "realise", "recognise"
]

australian_local_words = [
    "petrol", "mum", "uni", "footpath", "suburb",
    "holiday house", "bushfire", "bottle shop", "esky", "bloke"
]

us_signals = [
    "us congress", "senate", "white house", "washington",
    "new york", "california", "texas", "nasdaq", "wall street",
    "s&p 500", "mom", "gas station", "apartment", "downtown"
]

uk_signals = [
    ".co.uk", "bbc.co.uk", "theguardian.com/uk", "nhs",
    "downing street", "bank of england", "manchester",
    "london", "scotland", "wales", "mp"
]

# -------------------------
# Classification function
# -------------------------

def classify_australian_text(text_lower, url_lower):
    score = 0
    evidence = []

    # Level 1: strong domain signals
    for domain in australian_domains:
        if domain in url_lower:
            score += 5
            evidence.append(f"+5 Australian domain: {domain}")
            break

    # Level 1: Australian institutions, places, and market signals
    matched_keywords = [kw for kw in australian_keywords if kw in text_lower or kw in url_lower]
    if matched_keywords:
        keyword_score = min(len(matched_keywords), 4) * 2
        score += keyword_score
        evidence.append(f"+{keyword_score} Australian context keywords: {matched_keywords[:6]}")

    # Level 2: Australian spelling patterns
    matched_spellings = [w for w in australian_spellings if w in text_lower]
    if matched_spellings:
        spelling_score = min(len(matched_spellings), 3)
        score += spelling_score
        evidence.append(f"+{spelling_score} Australian/British spellings: {matched_spellings[:6]}")

    # Level 2: Australian local vocabulary
    matched_local_words = [w for w in australian_local_words if w in text_lower]
    if matched_local_words:
        local_word_score = min(len(matched_local_words), 3)
        score += local_word_score
        evidence.append(f"+{local_word_score} Australian local words: {matched_local_words[:6]}")

    # Level 3: strong American exclusion signals
    matched_us = [w for w in us_signals if w in text_lower or w in url_lower]
    if matched_us:
        us_score = min(len(matched_us), 3) * 2
        score -= us_score
        evidence.append(f"-{us_score} US signals: {matched_us[:6]}")

    # Level 3: strong British exclusion signals
    matched_uk = [w for w in uk_signals if w in text_lower or w in url_lower]
    if matched_uk:
        uk_score = min(len(matched_uk), 3) * 2
        score -= uk_score
        evidence.append(f"-{uk_score} UK signals: {matched_uk[:6]}")

    # Final classification
    if score >= 4:
        label = "Australian-local"
    elif score <= -2:
        label = "Non-Australian"
    else:
        label = "Uncertain"

    return {
        "label": label,
        "score": score,
        "evidence": evidence
    }

# -------------------------
# Run classification
# -------------------------

result = classify_australian_text(text_lower, url_lower)

print("Index:", row.name)
print("URL:", url)
print("Classification:", result["label"])
print("Score:", result["score"])
print("\nEvidence:")
for item in result["evidence"]:
    print("-", item)
```
