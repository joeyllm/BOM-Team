Ideas to differentiate Australian and Non-Australian texts :
- Date Format: if text contains a date, Australian: DD/MM/YYYY, American: MM/DD/YYY (but can only be used when the day > 12)
- Spelling Differences: non-Australian like organize, analyze, recognize, realize, color, honor
- Australian slang : arvo, servo, bottle-o, tradie, ute, footpath
- look at the url columns and look for Australian domains like .au, abc.net etc.
