Try to retrieve 1 hour data


```python
dataset = 'reanalysis-era5-single-levels'
request = {
    'product_type': 'reanalysis',
    'variable': ['10m_u_component_of_wind', '10m_v_component_of_wind'],
    'year': ['2024'],
    'month': ['03'],
    'day': ['01'],
    'time':['13:00'],
    'data_format': 'netcdf',
}
target = 'wind_2024_03_01_13.nc'

client.retrieve(dataset, request, target)
```

Retrieve 1 month data (February 2024), all the hourly data

```python
client.retrieve(
    'reanalysis-era5-single-levels',
    {
        'product_type': 'reanalysis',
        'variable': [
            '10m_u_component_of_wind',
            '10m_v_component_of_wind',
            '2m_temperature'
        ],
        'year': '2024',
        'month': '02',
        'day': [f'{d:02d}' for d in range(1, 32)],
        'time': [f'{h:02d}:00' for h in range(24)],  # all 24 hours
        'format': 'netcdf',
    },
    'era5_Feb2024.nc'
)
```

total data size of 1 month data hourly : 3.6 GB
